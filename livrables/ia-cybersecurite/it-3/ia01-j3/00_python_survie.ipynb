## 1. Variables et types

Une **variable** est un nom posé sur une valeur. Le signe `=` ne veut pas dire « égal », il veut dire « range cette valeur sous ce nom ». Tout ce qui suit un `#` est un **commentaire** : Python l'ignore, il est écrit pour les humains.

```python
ip = "203.0.113.51"     # du texte : une chaîne de caractères (str), toujours entre guillemets
echecs = 87             # un nombre entier (int)
octets_mo = 2.5         # un nombre à virgule (float) : on écrit un POINT, pas une virgule
externe = True          # un booléen (bool) : True ou False

print(ip, echecs, octets_mo, externe)
print(type(ip), type(echecs), type(octets_mo), type(externe))
```

## 6. pandas : un tableur dans Python

**pandas** charge un fichier CSV dans un tableau appelé **DataFrame**, que l'on nomme presque toujours `df`. Chaque ligne est une observation, chaque colonne une caractéristique. C'est le même format qu'au J1 : l'activité d'une adresse IP pendant une heure.

```python
df = pd.read_csv("hexalp_ssh_sept2026.csv")
df.head()             # les 5 premières lignes, comme `head -5`
```

```python
print(df.shape)       # (nombre de lignes, nombre de colonnes)
print(df.columns)     # le nom des colonnes
```

```python
df["echecs_auth"]             # une colonne
```

```python
df[["ip_source", "echecs_auth", "etiquette"]]     # plusieurs colonnes : une liste entre crochets
```

```python
print(df["echecs_auth"].max())              # la valeur maximale
print(df["origine"].value_counts())         # compter les valeurs, comme `sort | uniq -c`
```

```python
# Filtrer : garder les lignes où la condition est vraie, comme `grep` ou `awk '$5 > 50'`
df[df["echecs_auth"] > 50]
```

```python
# Trier, comme `sort -nr`, puis garder les 10 premières lignes
df.sort_values("echecs_auth", ascending=False).head(10)
```
